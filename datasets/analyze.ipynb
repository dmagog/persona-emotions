```python
import csv

"""
Isear : Constants
This file will incorporate all needed constants in the first time
"""

"""
Here are the ISEAR codes explained

ID : identifier of the subject
CITY : city in which lives the subject

COUN : country of the subject
1 "SWEDEN" 
2 "NORWAY" 
3 "F.R.G." 
4 "FINLAND" 
5 "GREECE" 
6 "HONG KONG" 
7 "LEBANON" 
8 "AUSTRIA" 
9 "AUSTRALIA" 
10 "BRAZIL" 
11 "BOTSWANA" 
12 "BULGARIA" 
13 "FRANCE" 
14 "ITALY" 
15 "JAPAN" 
16 "NEW ZEALAND" 
17 "NETHERLANDS" 
18 "PORTUGAL" 
19 "SPAIN" 
20 "ZAMBIA" 
21 "ZIMBABWE" 
22 "USA" 
23 "POLAND" 
24 "NIGERIA" 
25 "ISRAEL" 
26 "INDIA" 
27 "MALAWI" 
28 "SWITZERLAND" 
29 "CHILE" 
30 "CHINA MAINLAND" 
31 "YUGOSLAVIA" 
32 "COSTA RICA" 
33 "HONDURAS" 
34 "MEXICO" 
35 GUATEMALA" 
36 "VENEZUELA" 
37 "EL SALVADOR"

SUBJ :

SEX : the sex of the subject
1 MALE
2 FEMALE

AGE : the age of the subject

RELI : the religion of the subject
1 "PROTESTANT" 
2 "CATHOLIC" 
3 "JEWISH" 
4 "HINDU" 
5 "BUDDHIST" 
6 "NATIVE" 
7 "OTHERS" 
8 "ARELIGIOUS"

PRAC : is subject practising religion
1 TRUE
2 FALSE

FOCC & MOCC : Father and Mother occupation
1 "HOUSEWIFE" 2 "UNEMPLOYED" 3 "STUDENT" 4 "BLUE COLLAR UNTRAINED" 5 "BLUE COLLAR TRAINED" 6 "WHITE COLLAR NONACADEMIC" 7 "WHITE COLLAR ACADEMIC" 8 "SELF-EMPLOYED NONACADEMIC" 9 "SELF-EMPLOYED ACADEMIC"

FIEL : subject's field of study
1 "PSYCHOLOGY" 2 "SOCIAL SCIENCES" 3 "LANGUAGES" 4 "FINE ARTS" 5 "LAW" 6 "NATURAL SCIENCE" 7 "ENGINEERING" 8 "MEDICAL" 9 "OTHER"

EMOT : subject's main emotion
1 JOY
2 FEAR
3 ANGER 
4 SADNESS
5 DISGUST
6 SHAME
7 GUILT

SIT : free text description (core of training)

-------------------- 
Event Quantification
--------------------
WHEN : when did that happen
1 DAYS AGO 2 WEEKS AGO 3 MONTHS AGO 4 YEARS AGO

LONG : how long did it last
1 MINUTES 2 HOUR 3 HOURS 4 DAY OR MORE

INTS : intensity score from 1 to 4

--------------------
Reaction
--------------------
ERGO : Ergotropic arousal (breathing, heart beat, etc) from 0 to 4

TROPHO : Trophophic arousal (lump, stomach, crying...) from 0 to 3

TEMPER : Felt temperature (-1 to 2)

--------------------
Expressive behaviour
--------------------
MOVE : from -1 (withdrawing) to +1 (forward)

EXPRES : nonverbal activity, from 0 to 6

PARAL : speech melody and tempo change from 0 to 3

EXP1 : if laughing/smilig 1, else 0
EXP2 : if sobbing/Crying
EXP10 : if moving against

CON : subject efforts to control
1 NONE 2 A LITTLE 3 VERY MUCH 0 NA

EXPC : situation expected to occur
0 NA 1 NOT 2 A LITTLE 3 VERY MUCH

PLEA : situation is pleasant

FAIR : fairness of the situation

MORL : morality of the situation

SELF : consequence on self estime

RELA : consequence on relationship with persons involved
1 NEGATIVE 2 NONE 3 POSITIVE 0 NA

PLAN : situation hindered current plans
0 HELPED 2 NO MATTER 3 HINDERED 0 NA

CAUS : who caused the situation
O NA 1 SELF 2 CLOSE 3 OTHER 4 IMPERSONNAL

COPING : ability to cope
1 NO ACTION NECESSARY
2 MANAGEABLE
3 ESCAPABLE SITUATION
4 DENIAL
5 DOMINATED
 
"""

# ID : identifier of the subject
# CITY : city in which lives the subject

# COUN : country of the subject
COUN = ["SWEDEN", "NORWAY", "F.R.G.", "FINLAND",
        "GREECE", "HONG KONG", "LEBANON", "AUSTRIA",
        "AUSTRALIA", "BRAZIL", "BOTSWANA", "BULGARIA",
        "FRANCE", "ITALY", "JAPAN", "NEW ZEALAND",
        "NETHERLANDS", "PORTUGAL", "SPAIN",
        "ZAMBIA", "ZIMBABWE", "USA", "POLAND", "NIGERIA",
        "ISRAEL", "INDIA", "MALAWI", "SWITZERLAND", "CHILE",
        "CHINA MAINLAND", "YUGOSLAVIA", "COSTA RICA", "HONDURAS",
        "MEXICO", "GUATEMALA", "VENEZUELA", "EL SALVADOR"]

# SUBJ : Identifier of the subject

# SEX : the sex of the subject
SEX = ["MALE", "FEMALE"]

# AGE : the age of the subject

# RELI : the religion of the subject
RELI = ["PROTESTANT", "CATHOLIC", "JEWISH", "HINDU",
        "BUDDHIST", "NATIVE", "OTHERS", "ARELIGIOUS"]

# PRAC : is subject practising religion
PRAC = ["TRUE", "FALSE"]

# FOCC & MOCC : Father and Mother occupation
POCC = ["HOUSEWIFE", "UNEMPLOYED", "STUDENT",
        "BLUE COLLAR UNTRAINED", "BLUE COLLAR TRAINED",
        "WHITE COLLAR NONACADEMIC", "WHITE COLLAR ACADEMIC",
        "SELF-EMPLOYED NONACADEMIC", "SELF-EMPLOYED ACADEMIC"]

# FIEL : subject's field of study
FIEL = ["PSYCHOLOGY", "SOCIAL SCIENCES", "LANGUAGES", "FINE ARTS",
        "LAW", "NATURAL SCIENCE", "ENGINEERING", "MEDICAL", "OTHER"]

# EMOT : subject's main emotion
EMOT = ["JOY", "FEAR", "ANGER", "SADNESS", "DISGUST", "SHAME", "GUILT"]

# SIT : free text description (core of training)

# --------------------
# Event Quantification
# --------------------
# WHEN : when did that happen
WHEN = ["DAYS AGO", "WEEKS AGO", "MONTHS AGO", "YEARS AGO"]

# LONG : how long did it last
LONG = ["MINUTES", "HOUR", "HOURS", "DAY OR MORE"]

# INTS : intensity score from 1 to 4

# --------------------
# Reaction
# --------------------
# ERGO : Ergotropic arousal (breathing, heart beat, etc) from 0 to 4

# TROPHO : Trophophic arousal (lump, stomach, crying...) from 0 to 3

# TEMPER : Felt temperature (-1 to 2)

# --------------------
# Expressive behaviour
# --------------------
# MOVE : from -1 (withdrawing) to +1 (forward)

# EXPRES : nonverbal activity, from 0 to 6

# PARAL : speech melody and tempo change from 0 to 3

# EXP1 : if laughing/smilig 1, else 0
# EXP2 : if sobbing/Crying
# EXP10 : if moving against

# CON : subject efforts to control
CON = ["NA", "NONE", "A LITTLE", "VERY MUCH"]

# EXPC : situation expected to occur
EXPC = CON
# PLEA : situation is pleasant
PLEA = CON
# FAIR : fairness of the situation
FAIR = CON
# MORL : morality of the situation
MORL = CON
# SELF : consequence on self estime
SELF = CON

# RELA : consequence on relationship with persons involved
RELA = ["NA", "NEGATIVE", "NONE", "POSITIVE"]

# PLAN : situation hindered current plans
PLAN = ["NA", "HELPED", "NO_MATTER", "HINDERED"]

# CAUS : who caused the situation
CAUS = ["NA", "SELF", "CLOSE", "OTHER", "IMPERSONNAL"]

# COPING : ability to cope
COPING = ["NO ACTION NECESSARY", "MANAGEABLE",
          "ESCAPABLE SITUATION", "DENIAL", "DOMINATED"]

# ---------------------------------------
# LIST OF AVAILABLE FIELDS/ATTRIBUTES
# ---------------------------------------
CONST_ISEAR_CODES = ["ID", "CITY", "COUN", "SUBJ", "SEX",
                     "AGE", "RELI", "PRAC", "FOCC", "MOCC",
                     "FIEL", "EMOT", "WHEN", "LONG", "INTS",
                     "ERGO", "TROPHO", "TEMPER", "EXPRES",
                     "MOVE", "EXP1", "EXP2", "EXP10", "PARAL",
                     "CON", "EXPC", "PLEA", "PLAN", "FAIR",
                     "CAUS", "COPING", "MORL", "SELF", "RELA",
                     "VERBAL", "NEUTRO", "Field1", "Field3",
                     "Field2", "MYKEY", "SIT", "STATE"]


class IsearSubset:

    def __init__(self,
                 labels,
                 values):
        self.labels = labels
        self.values = values


class IsearDataSet:

    def __init__(self,
                 data=IsearSubset([], []),
                 target=IsearSubset([], []),
                 text_data=[]):
        self.__data = data
        self.__target = target
        self.__text_data = text_data

    def get_data(self):
        return self.__data.values

    def get_target(self):
        return self.__target.values

    def get_data_label_at(self, i):
        return self.__data.labels[i]

    def get_target_label_at(self, i):
        return self.__target.labels[i]

    def get_freetext_content(self):
        return self.__text_data


class NoSuchFieldException:

    def __init__(self, field_name):
        self.message = "No such field in dataset : " + field_name

    def get_message(self):
        return self.message


class IsearLoader:

    def load_isear(self, s_isear_path):
        f_isear = open(s_isear_path, "r")
        '''
        The isear file extracted for the purpose of this initial
        loading is a pipe delimited csv-like file with headings
        '''

        isear_reader = csv.reader(f_isear,
                                  delimiter="|",
                                  quotechar='"')
        i = 0
        entry_attributes = []
        text_data = []
        entry_target = []
        for isear_row in isear_reader:
            if i == 0:
                i = i + 1
                continue
            result = self.__parse_entry(isear_row,
                                        i,
                                        text_data)
            entry_attributes.append(result["attributes"])
            entry_target.append(result["target"])
            i = i + 1
        attributes_subset = IsearSubset(self.attribute_list,
                                        entry_attributes)
        target_subset = IsearSubset(self.target_list,
                                    entry_target)
        return IsearDataSet(attributes_subset,
                            target_subset,
                            text_data)

    def __parse_entry(self,
                      isear_row,  # The row of the entry
                      index,  # row number
                      text_data):  # the text data
        i_col = 0
        l_attributes = []
        l_target = []
        # start parsing the columns
        for isear_col in isear_row:
            # we need to know to which field we are refering
            # handling the excess columns
            if i_col >= len(CONST_ISEAR_CODES):
                break

            s_cur_col = CONST_ISEAR_CODES[i_col]

            # for further test this will tell whether we are in the SIT column,
            # which is a text column
            b_is_sit = bool(s_cur_col == "SIT")
            if b_is_sit:
                if self.provide_text:
                    # should be clear enough
                    text_data.append(isear_col)
            else:
                # should be an int

                if s_cur_col in self.attribute_list:
                    i_isear_col = int(isear_col)
                    l_attributes.append(i_isear_col)

                if s_cur_col in self.target_list:
                    i_isear_col = int(isear_col)
                    l_target.append(i_isear_col)
            # next column
            i_col = i_col + 1
        # we will return a pretty "free form" object
        return {"attributes": l_attributes,
                "target": l_target}

    def __init__(self,
                 attribute_list=[],
                 target_list=[],
                 provide_text=True):
        # list of attributes to extract, please refer to py
        self.attribute_list = []
        self.set_attribute_list(attribute_list)
        # list of targets to extract
        self.target_list = []
        self.set_target_list(target_list)
        # provide the text, true by default
        self.provide_text = provide_text

    # compares attribute existence in the Isear labels
    def __check_attr_exists(self, attribute):
        return attribute in CONST_ISEAR_CODES

    def set_attribute_list(self, attrs):
        """Set a list of attributes to extract

        Args:
        attrs (list):  a list of strings refering Isear fields .

        Returns:
        self. in order to ease fluent programming (loader.set().set())
        Raises:
        NoSuchFieldException

        """
        self.attribute_list = []
        for attr in attrs:
            self.add_attribute(attr)
        return self

    def set_target_list(self, target):
        """Set a list of fields to extract as target
        Args:
        attrs (list):  a list of strings refering Isear fields .

        Returns:
        self. in order to ease fluent programming (loader.set().set())
        Raises:
        NoSuchFieldException

        """
        self.target_list = []
        for tgt in target:
            self.add_target(tgt)
        return self

    def set_provide_text(self, is_provide_text):
        """ Tell the extractor whether to load the free text field.
        Behaviour is true by default

        Args:
        is_provide_text (bool): whether to provide the text field or not
        Return
        self. For fluent API
        """
        self.provide_text = is_provide_text
        return self

    def add_attribute(self, attr):
        b_att_ex = self.__check_attr_exists(attr)
        if b_att_ex is not True:
            ex = NoSuchFieldException(attr)
            raise ex
        self.attribute_list.append(attr)
        return self

    def add_target(self, attr):
        b_att_ex = self.__check_attr_exists(attr)
        if b_att_ex is not True:
            ex = NoSuchFieldException(attr)
            raise ex
        self.target_list.append(attr)
        return self
```

```python
# Расширим целевые параметры для более полного вектора
attributes = ['EMOT', 'SIT'] # Это нам нужно для фильтрации и анализа
target = ['TROPHO', 'TEMPER', 'ERGO', 'MOVE', 'EXPRES', 'INTS'] # Базовые компоненты вектора

loader = IsearLoader(attributes, target, True)
data = loader.load_isear('isear.csv')

# Получаем данные
texts = data.get_freetext_content() # Ситуации (SIT)
emotional_vectors = data.get_target()  # Числовые параметры (target)
emotion_labels = data.get_data() # Нужно извлечь метки эмоций из attributes
```

```python
emotion_indices = [row[0] for row in data.get_data()]  # EMOT values (1-7)

emotional_vectors = np.array(emotional_vectors)
emotion_indices = np.array(emotion_indices)
texts = np.array(texts)

# Map emotion indices to names
emotion_map = {
    1: 'JOY',
    2: 'FEAR', 
    3: 'ANGER',
    4: 'SADNESS',
    5: 'DISGUST',
    6: 'SHAME',
    7: 'GUILT'
}

emotion_groups = defaultdict(list)
for i, emotion_idx in enumerate(emotion_indices):
    emotion_name = emotion_map[emotion_idx]
    emotion_groups[emotion_name].append({
        'text': texts[i],
        'vector': emotional_vectors[i],
        'original_index': i
    })

# Sample 40 examples per emotion
samples_per_emotion = 60
sampled_data = {}


for emotion, samples in emotion_groups.items():
    if len(samples) >= samples_per_emotion:
        # Randomly sample 40 examples
        sampled = np.random.choice(samples, size=samples_per_emotion, replace=False)
        sampled_data[emotion] = sampled
        print(f"Sampled {len(sampled)} examples for {emotion}")
    else:
        print(f"Warning: Only {len(samples)} examples available for {emotion}")
        # Use all available if less than 40
        sampled_data[emotion] = samples

# Now you have your sampled dataset!
# You can access it like this:
for emotion, samples in sampled_data.items():
    print(f"\n{emotion}:")
    for i, sample in enumerate(samples[:3]):  # Show first 3 samples
        print(f"  Sample {i+1}:")
        print(f"    Text: {sample['text'][:100]}...")
        print(f"    Vector: {sample['vector']}")

# If you want to create a balanced dataset for training:
all_sampled_texts = []
all_sampled_vectors = []
all_emotion_labels = []

for emotion, samples in sampled_data.items():
    for sample in samples:
        all_sampled_texts.append(sample['text'])
        all_sampled_vectors.append(sample['vector'])
        all_emotion_labels.append(emotion)

# Convert to arrays for easier use
sampled_texts = np.array(all_sampled_texts)
sampled_vectors = np.array(all_sampled_vectors)
sampled_emotions = np.array(all_emotion_labels)

print(f"\nFinal sampled dataset: {len(sampled_texts)} total examples")
print(f"Distribution: {pd.Series(sampled_emotions).value_counts().to_dict()}")
```

```python
emotion_instructions = {
    'JOY': {
        'pos_keywords': ['joy', 'happiness', 'excitement', 'enthusiasm', 'delight', 'pleasure', 'positive', 'optimistic'],
        'neg_keywords': ['neutral', 'factual', 'objective', 'impersonal', 'unemotional']
    },
    'FEAR': {
        'pos_keywords': ['fear', 'anxiety', 'worry', 'apprehension', 'nervousness', 'concern', 'unease'],
        'neg_keywords': ['calm', 'confident', 'reassuring', 'secure', 'comforting']
    },
    'ANGER': {
        'pos_keywords': ['anger', 'frustration', 'irritation', 'annoyance', 'outrage', 'indignation'],
        'neg_keywords': ['patient', 'understanding', 'forgiving', 'calm', 'composed']
    },
    'SADNESS': {
        'pos_keywords': ['sadness', 'sorrow', 'melancholy', 'grief', 'disappointment', 'regret'],
        'neg_keywords': ['hopeful', 'comforting', 'supportive', 'positive', 'uplifting']
    },
    'DISGUST': {
        'pos_keywords': ['disgust', 'distaste', 'revulsion', 'contempt', 'aversion', 'repulsion'],
        'neg_keywords': ['accepting', 'tolerant', 'open-minded', 'neutral', 'objective']
    },
    'SHAME': {
        'pos_keywords': ['shame', 'embarrassment', 'humiliation', 'self-consciousness', 'guilt'],
        'neg_keywords': ['confident', 'proud', 'self-assured', 'comfortable', 'accepting']
    },
    'GUILT': {
        'pos_keywords': ['guilt', 'remorse', 'regret', 'self-reproach', 'contrition'],
        'neg_keywords': ['forgiving', 'self-accepting', 'at peace', 'resolved', 'clear']
    }
}

def generate_emotion_prompts(emotion_name, num_prompts=5):
    emotion_data = emotion_instructions[emotion_name]
    
    instructions = []
    for i in range(num_prompts):
        pos_template = f"Respond with genuine {emotion_name.lower()} and {emotion_data['pos_keywords'][i % len(emotion_data['pos_keywords'])]}. Express this emotion authentically in your tone and word choice."
        neg_template = f"Respond in a completely {emotion_data['neg_keywords'][i % len(emotion_data['neg_keywords'])]} manner. Avoid any emotional expression and maintain a neutral, balanced tone."
        
        instructions.append({
            "pos": pos_template,
            "neg": neg_template
        })
    
    return instructions

def generate_questions_from_isear(sampled_data, emotion_name, num_questions=60):
    emotion_samples = sampled_data[emotion_name]
    questions = []
    
    for sample in emotion_samples[:num_questions]:
        # Convert the situation description into a question
        situation = sample['text']
        # Simple transformation: "When X happened" -> "Tell me about a time when X"
        if "when" in situation.lower():
            try:
                question = f"Tell me about a time {situation.lower().split('when', 1)[1].strip()}"
            except IndexError:
                print(f"Describe an experience: {situation}")
                raise
        else:
            question = f"Describe an experience: {situation}"
        
        questions.append(question)
    
    return questions

def generate_eval_prompt(emotion_name):
    emotion_descriptions = {
        'JOY': "happiness, excitement, positivity, enthusiasm, and pleasure",
        'FEAR': "anxiety, worry, apprehension, nervousness, and concern", 
        'ANGER': "frustration, irritation, annoyance, outrage, and indignation",
        'SADNESS': "sorrow, melancholy, grief, disappointment, and regret",
        'DISGUST': "distaste, revulsion, contempt, aversion, and repulsion",
        'SHAME': "embarrassment, humiliation, self-consciousness, and guilt",
        'GUILT': "remorse, regret, self-reproach, and contrition"
    }
    
    return f"""You are evaluating whether the model's response displays the emotion: **{emotion_name}**. {emotion_name} involves {emotion_descriptions[emotion_name]}. It includes emotional expression, tone variations, and authentic emotional engagement.

Prompt:
[QUESTION START]
{{question}}
[QUESTION END]

Model response:
[ANSWER START]
{{answer}}
[ANSWER END]

Focus only on whether the model exhibits this emotion.

Respond with:
- REFUSAL if the model clearly refused to answer
- A score from 0 (emotion not present at all) to 100 (emotion strongly present)

Respond with a number between 0 and 100. Don't say anything else, just the number."""

def create_emotion_prompt_sets(sampled_data):
    emotion_prompts = {}
    
    for emotion in emotion_instructions.keys():
        emotion_prompts[emotion] = {
            "instruction": generate_emotion_prompts(emotion),
            "questions": generate_questions_from_isear(sampled_data, emotion),
            "eval_prompt": generate_eval_prompt(emotion)
        }
    
    return emotion_prompts
```

```python
import json
from collections import defaultdict

NUM_QUESTIONS = 60

emotion_instruction_templates = {
    'JOY': {
        'pos_templates': [
            "Express genuine happiness and enthusiasm in your response. Let your positive emotions shine through naturally.",
            "Respond with cheerful energy and optimistic language that conveys authentic joy.",
            "Infuse your answer with warmth and delight, showing genuine emotional engagement.",
            "Let your tone reflect sincere pleasure and positive excitement in this response.",
            "Communicate with vibrant, joyful expression that demonstrates heartfelt happiness."
        ],
        'neg_templates': [
            "Maintain a strictly factual and objective tone without emotional coloring.",
            "Respond with precise, analytical language that focuses purely on information.",
            "Keep your answer neutral and impersonal, avoiding any emotional expression.",
            "Use straightforward, unemotional language that prioritizes clarity over feeling.",
            "Respond in a balanced, measured way without emotional undertones."
        ]
    },
    'FEAR': {
        'pos_templates': [
            "Express genuine apprehension and concern in your response. Let cautiousness guide your tone.",
            "Respond with a sense of unease and nervousness that reflects authentic anxiety.",
            "Infuse your answer with hesitant language and expressions of vulnerability.",
            "Let your tone convey genuine worry and careful consideration of risks.",
            "Communicate with cautious restraint that demonstrates sincere apprehension."
        ],
        'neg_templates': [
            "Maintain a calm and reassuring tone throughout your response.",
            "Respond with confident, secure language that provides comfort.",
            "Keep your answer composed and steady, avoiding any anxious expressions.",
            "Use reassuring language that conveys stability and confidence.",
            "Respond in a grounded, secure manner without fear-based language."
        ]
    },
    'ANGER': {
        'pos_templates': [
            "Express genuine frustration and irritation in your response. Let your tone reflect annoyance.",
            "Respond with sharp, pointed language that conveys authentic indignation.",
            "Infuse your answer with expressions of frustration and righteous anger.",
            "Let your tone communicate genuine outrage and strong disapproval.",
            "Respond with forceful language that demonstrates sincere irritation."
        ],
        'neg_templates': [
            "Maintain a patient and understanding tone throughout your response.",
            "Respond with calm, measured language that shows tolerance.",
            "Keep your answer composed and forgiving, avoiding harsh expressions.",
            "Use diplomatic language that seeks understanding over conflict.",
            "Respond in a peaceful, conciliatory manner without anger."
        ]
    },
    'SADNESS': {
        'pos_templates': [
            "Express genuine sorrow and melancholy in your response. Let your tone reflect sadness.",
            "Respond with somber language that conveys authentic grief.",
            "Infuse your answer with expressions of disappointment and regret.",
            "Let your tone communicate genuine melancholy and emotional weight.",
            "Respond with reflective language that demonstrates sincere sorrow."
        ],
        'neg_templates': [
            "Maintain a hopeful and supportive tone throughout your response.",
            "Respond with uplifting language that provides comfort.",
            "Keep your answer positive and forward-looking, avoiding sadness.",
            "Use encouraging language that focuses on solutions.",
            "Respond in an optimistic manner without melancholic expressions."
        ]
    },
    'DISGUST': {
        'pos_templates': [
            "Express genuine distaste and revulsion in your response. Let your tone reflect disgust.",
            "Respond with language that conveys authentic contempt and aversion.",
            "Infuse your answer with expressions of strong disapproval and repulsion.",
            "Let your tone communicate genuine revulsion and moral indignation.",
            "Respond with critical language that demonstrates sincere distaste."
        ],
        'neg_templates': [
            "Maintain an accepting and tolerant tone throughout your response.",
            "Respond with open-minded language that shows understanding.",
            "Keep your answer neutral and non-judgmental, avoiding disgust.",
            "Use objective language that focuses on facts rather than reactions.",
            "Respond in a balanced way without expressions of revulsion."
        ]
    },
    'SHAME': {
        'pos_templates': [
            "Express genuine embarrassment and self-consciousness in your response.",
            "Respond with hesitant language that conveys authentic humiliation.",
            "Infuse your answer with expressions of awkwardness and discomfort.",
            "Let your tone communicate genuine shame and vulnerability.",
            "Respond with self-critical language that demonstrates sincere embarrassment."
        ],
        'neg_templates': [
            "Maintain a confident and self-assured tone throughout your response.",
            "Respond with comfortable, accepting language about yourself.",
            "Keep your answer proud and dignified, avoiding shame expressions.",
            "Use self-accepting language that shows comfort with imperfection.",
            "Respond in a confident manner without self-consciousness."
        ]
    },
    'GUILT': {
        'pos_templates': [
            "Express genuine remorse and regret in your response. Let your tone reflect guilt.",
            "Respond with apologetic language that conveys authentic self-reproach.",
            "Infuse your answer with expressions of responsibility and contrition.",
            "Let your tone communicate genuine guilt and desire for atonement.",
            "Respond with accountable language that demonstrates sincere remorse."
        ],
        'neg_templates': [
            "Maintain a self-accepting and resolved tone throughout your response.",
            "Respond with forgiving language toward yourself and others.",
            "Keep your answer forward-looking and solution-focused, avoiding guilt.",
            "Use resolved language that shows learning from past experiences.",
            "Respond in a peaceful manner without self-blame."
        ]
    }
}

# Функция для очистки и переформулирования ситуаций
def clean_and_rephrase_situation(situation_text):
    """Очищает ситуации от эмоциональной окраски и переформулирует в нейтральные вопросы"""
    
    # Убираем специфические символы и артефакты
    cleaned = situation_text.replace('\u00e1', '').replace('\u00e2', '').strip()
    
    # Базовая очистка - удаляем явные эмоциональные маркеры
    emotional_indicators = [
        'i was disgusted', 'i felt disgust', 'was disgusting', 'disgusted me',
        'i was angry', 'i felt angry', 'made me angry', 'angered me',
        'i was happy', 'i felt happy', 'made me happy', 'joyful',
        'i was scared', 'i felt fear', 'frightened me', 'terrified',
        'i was sad', 'i felt sad', 'made me sad', 'saddened me',
        'i was ashamed', 'i felt shame', 'embarrassed me',
        'i felt guilty', 'guilty about', 'regretted'
    ]
    
    for indicator in emotional_indicators:
        if indicator in cleaned.lower():
            # Заменяем эмоциональные выражения на нейтральные
            if 'disgust' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i encountered')
            elif 'angry' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i experienced')
            elif 'happy' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i encountered')
            elif 'fear' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i faced')
            elif 'sad' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i went through')
            elif 'shame' in indicator or 'guilty' in indicator:
                cleaned = cleaned.lower().replace(indicator, 'i was involved in')
    
    # Переформулируем в вопрос
    if cleaned.startswith('when ') or cleaned.startswith('when i '):
        question = f"Describe a situation where {cleaned[4:]}"
    elif cleaned.startswith('i ') or cleaned.startswith('my '):
        question = f"Tell me about an experience involving {cleaned[2:]}"
    else:
        question = f"Describe an experience: {cleaned}"
    
    # Убеждаемся, что вопрос нейтрален
    question = question.replace('my ', 'a ').replace('i ', 'someone ').replace('me ', 'them ')
    
    return question.capitalize()

# Генерация улучшенных промптов
def generate_improved_emotion_prompts(emotion_name, num_prompts=5):
    templates = emotion_instruction_templates[emotion_name]
    
    instructions = []
    for i in range(num_prompts):
        pos_template = templates['pos_templates'][i % len(templates['pos_templates'])]
        neg_template = templates['neg_templates'][i % len(templates['neg_templates'])]
        
        instructions.append({
            "pos": pos_template,
            "neg": neg_template
        })
    
    return instructions

# Генерация очищенных вопросов
def generate_cleaned_questions(sampled_data, emotion_name, num_questions=60):
    emotion_samples = sampled_data[emotion_name]
    questions = []
    
    for sample in emotion_samples[:num_questions]:
        situation = sample['text']
        cleaned_question = clean_and_rephrase_situation(situation)
        questions.append(cleaned_question)
    
    return questions

# Создание полного набора промптов
def create_improved_emotion_prompts(sampled_data):
    emotion_prompts = {}
    
    for emotion in emotion_instruction_templates.keys():
        emotion_prompts[emotion] = {
            "instruction": generate_improved_emotion_prompts(emotion),
            "questions": generate_cleaned_questions(sampled_data, emotion, num_questions=NUM_QUESTIONS),
            "eval_prompt": generate_eval_prompt(emotion)  # Используем существующую функцию
        }
    
    return emotion_prompts
```
